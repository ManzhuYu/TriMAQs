```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

import geopandas as gpd
from shapely import wkt
from shapely.geometry import box
from shapely.geometry import Point
from shapely.ops import nearest_points
from scipy.spatial import cKDTree
from sklearn.impute import SimpleImputer

from osgeo import gdal
from scipy.interpolate import griddata
from scipy.spatial import distance
from scipy.spatial import cKDTree
from pyproj import CRS, Proj, Transformer
from shapely.geometry import box

from datetime import datetime, timedelta

import xarray as xr
import os
from tqdm import tqdm

import warnings
warnings.filterwarnings("ignore")
```

```python
def mean_with_nan(arrays):
    """
    Calculate the mean of multiple 2D arrays considering NaN values.
    If one element is NaN and others have values, the result takes the non-NaN value(s).
    """
    
    stacked_arrays = np.stack(arrays)    
    
    sum_array = np.nansum(stacked_arrays, axis=0)   
    
    count_non_nan = np.sum(~np.isnan(stacked_arrays), axis=0).astype(float)   
    
    count_non_nan[count_non_nan == 0] = np.nan    
    
    mean_array = sum_array / count_non_nan
    
    return mean_array
```

```python
grid = pd.read_csv('grid_proj_1km.csv')[['x', 'y', 'geometry', 'lon', 'lat']]

grid['geometry'] = grid['geometry'].apply(wkt.loads)

grid_gdf = gpd.GeoDataFrame(grid, geometry='geometry', crs='ESRI:102003')
```

# Batch process TROPOMI NO2 to 1km grids

```python
tropomi_no2_path = 'TROPOMI NO2/'

files = [f for f in os.listdir('TROPOMI_vs_NO2/') if f.endswith('csv')]
files.sort()
```

```python
for f in files:

    date = f.split('.')[0]
    print(date)

    if not os.path.exists('TROPOMI_vs_NO2/1km_grid/'+date+'.csv'):
    
        year, month, day = date[:4], date[5:7], date[8:]
        
        tropomi_no2_files = [f for f in os.listdir(tropomi_no2_path) if '__'+date in f]
        
        no2_grids_list = []
        
        if len(tropomi_no2_files) >= 1:
        
            for tropomi_no2_file in tropomi_no2_files:
            
                ds = xr.open_dataset(tropomi_no2_path + tropomi_no2_file,
                                group='PRODUCT')
                
                ds = ds.rio.write_crs(4326)
                
                obs_time = ds.time_utc.values[0][0]
                
                lon = ds['longitude'].rio.interpolate_na()[0,:,:]
                lat = ds['latitude'].rio.interpolate_na()[0,:,:]

                no2_converted = ds['nitrogendioxide_tropospheric_column'] * ds['nitrogendioxide_tropospheric_column'].multiplication_factor_to_convert_to_molecules_percm2
                
                no2_converted = no2_converted.where(ds['qa_value']>0.5, np.nan)
                
                no2_converted = no2_converted.where(no2_converted >= 0, np.nan)
                
                no2_converted = no2_converted[0,:,:]
        
                flat_lats = no2_converted['latitude'].values.flatten()
                flat_lons = no2_converted['longitude'].values.flatten()
                flat_no2 = no2_converted.values.flatten()
                
                mask = ~np.isnan(flat_no2)
                
                masked_lats = flat_lats[mask]
                masked_lons = flat_lons[mask]
                masked_no2 = flat_no2[mask]
                
                lat_da = xr.DataArray(masked_lats, dims='points')
                lon_da = xr.DataArray(masked_lons, dims='points')
                no2_da = xr.DataArray(masked_no2, dims='points', coords={'lat': lat_da, 'lon': lon_da})
                
                tree = cKDTree(np.c_[masked_lats, masked_lons])
                
                query_points = np.c_[grid_gdf['lat'].values, grid_gdf['lon'].values]
                
                distances, indices = tree.query(query_points, k=1)
                
                selected_no2_values = np.full_like(grid_gdf['lat'].values, np.nan, dtype=np.float64)
                
                valid_mask = distances < 0.05
                
                valid_indices = indices[valid_mask]
                
                selected_no2_values[valid_mask] = masked_no2[valid_indices]
                
                no2_grids_list.append(selected_no2_values)
    
        if len(no2_grids_list) > 0:
        
            grid_gdf['no2'] = mean_with_nan(no2_grids_list)
            
            grid_gdf[['no2']].to_csv('TROPOMI_vs_NO2/1km_grid/'+date+'.csv', index=False)
    
        else:
            print('no data')
```
